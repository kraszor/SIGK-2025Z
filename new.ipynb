```python
!pip install bvhio matplotlib torch numpy imageio

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import bvhio
import os
import math
import random


DATASET_ROOT = '/content/drive/MyDrive/SIGK/sigk5/'

FRAMES = 48
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BATCH_SIZE = 64
TIMESTEPS = 500
LR = 1e-4
EPOCHS = 5000
JOINTS = 9
TARGET_JOINTS = ['Hips', 'LeftUpLeg', 'LeftLeg', 'RightUpLeg', 'RightLeg', 'LeftArm', 'LeftForeArm', 'RightArm', 'RightForeArm']

CONNECTIONS = [
    (0, 1), (1, 2),
    (0, 3), (3, 4),
    (0, 5), (5, 6),
    (0, 7), (7, 8)
]
```

```python
class RealMotionDataset(torch.utils.data.Dataset):
    def __init__(self, root_dir):
        self.samples = []

        if not os.path.exists(root_dir):
            raise FileNotFoundError(f"Katalog {root_dir} nie istnieje. Wgraj pliki!")

        walk_set = self._load_category(os.path.join(root_dir, 'walk'), 0)
        jump_set = self._load_category(os.path.join(root_dir, 'jump'), 1)

        for m in walk_set:
            for _ in range(100):
                noise = np.random.randn(*m.shape) * 0.01
                self.samples.append((m + noise, 0))

        for m in jump_set:
            for _ in range(50):
                noise = np.random.randn(*m.shape) * 0.01
                self.samples.append((m + noise, 1))

    def _load_category(self, path, label):
        if not os.path.exists(path):
            print(f"Ostrzeżenie: Folder {path} nie istnieje.")
            return

        files = [f for f in os.listdir(path) if f.endswith('.bvh')]
        print(f"Znaleziono {len(files)} plików w {path}...")
        motion_set = []
        for filename in files:
            file_path = os.path.join(path, filename)
            try:
                motion = self._parse_bvh(file_path, label)
                motion_set.append(motion)
            except Exception as e:
                print(f"Błąd przy pliku {filename}: {e}")
        return motion_set

    def _parse_bvh(self, path, label):
        root = bvhio.readAsHierarchy(path)
        bvh = bvhio.readAsBvh(path)
        total_frames = bvh.FrameCount

        layout = root.layout()
        all_joints = [j for j, _, _ in layout]

        selected_nodes = []

        for name in TARGET_JOINTS:
            found = next((n for n in all_joints if name in n.Name), None)
            if not found and "UpLeg" in name:
                alt = name.replace("UpLeg", "Hip")
                found = next((n for n in all_joints if alt in n.Name), None)

            if not found: found = root
            selected_nodes.append(found)

        frames_data = []
        num_frames = len(root.Keyframes)
        step = max(1, num_frames // FRAMES)

        for i in range(0, num_frames, step):
            if len(frames_data) >= FRAMES: break
            root.loadPose(i)
            pose = [[n.PositionWorld.x, n.PositionWorld.y, n.PositionWorld.z] for n in selected_nodes]
            frames_data.append(pose)

        full_motion = np.array(frames_data)
        motion = full_motion - full_motion[:, 0:1, :]
        scale = np.max(np.abs(motion))
        if scale > 0: motion /= scale
        while motion.shape[0] < FRAMES:
            motion = np.concatenate([motion, motion[-1:]], axis=0)

        return motion[:FRAMES].transpose(2, 0, 1)


    def __len__(self):
        return len(self.sample)

    def __getitem__(self, idx):
        return torch.tensor(self.samples[idx][0], dtype=torch.float32), torch.tensor(self.samples[idx][1])


dataset = RealMotionDataset(DATASET_ROOT)
dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)
```

```python
# class SinusoidalPositionEmbeddings(nn.Module):
#     def __init__(self, dim):
#         super().__init__()
#         self.dim = dim

#     def forward(self, time):
#         device = time.device
#         half_dim = self.dim // 2
#         embeddings = math.log(10000) / (half_dim - 1)
#         embeddings = torch.exp(torch.arange(half_dim, device=device) * -embeddings)
#         embeddings = time[:, None] * embeddings[None, :]
#         embeddings = torch.cat((embeddings.sin(), embeddings.cos()), dim=-1)
#         return embeddings
def cosine_beta_schedule(timesteps, s=0.008):
    """
    cosine schedule as proposed in https://arxiv.org/abs/2102.09672
    """
    steps = timesteps + 1
    x = torch.linspace(0, timesteps, steps)
    alphas_cumprod = torch.cos(((x / timesteps) + s) / (1 + s) * math.pi * 0.5) ** 2
    alphas_cumprod = alphas_cumprod / alphas_cumprod[0]
    betas = 1 - (alphas_cumprod[1:] / alphas_cumprod[:-1])
    return torch.clip(betas, 0.0001, 0.9999).to(DEVICE)


class DiffusionUNet(nn.Module):
    def __init__(self, in_channels, out_channels, time_dim=64):
        super().__init__()
        self.time_mlp = nn.Sequential(
            # SinusoidalPositionEmbeddings(time_dim),
            nn.Linear(time_dim, out_channels),
            # nn.ReLU()
        )
        self.block1 = nn.Sequential(
            nn.Conv1d(in_channels, out_channels, 3, padding=1),
            nn.GroupNorm(8, out_channels),
            nn.SiLU(),
        )

        self.block2 = nn.Sequential(
            nn.Conv1d(out_channels, out_channels, 3, padding=1),
            nn.GroupNorm(8, out_channels),
            nn.SiLU(),
        )
        self.residual_conv = nn.Conv1d(in_channels, out_channels, 1) if in_channels != out_channels else nn.Identity()

    def forward(self, x, t):
        h = self.block1(x)
        time_emb = self.time_mlp(t).unsqueeze(-1)
        h = h + time_emb

        h = self.block2(h)
        return h + self.residual_conv(x)

class FullDiffusionModel(nn.Module):
    def __init__(self, num_joints, num_classes=2, hidden_dim=128):
        super().__init__()
        in_channels = 3 * num_joints

        self.time_mlp = nn.Sequential(
            nn.Linear(1, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
        )
        self.class_emb = nn.Embedding(num_classes, hidden_dim)

        self.inc = nn.Conv1d(in_channels, hidden_dim, 3, padding=1)
        self.down1 = DiffusionUNet(hidden_dim, hidden_dim*2, hidden_dim)
        self.down2 = DiffusionUNet(hidden_dim*2, hidden_dim*4, hidden_dim)

        self.mid1 = DiffusionUNet(hidden_dim*4, hidden_dim*4, hidden_dim)
        self.mid2 = DiffusionUNet(hidden_dim*4, hidden_dim*4, hidden_dim)

        self.up1 = DiffusionUNet(hidden_dim*8, hidden_dim*2, hidden_dim)
        self.up2 = DiffusionUNet(hidden_dim*4, hidden_dim, hidden_dim)

        self.outc = nn.Conv1d(hidden_dim, in_channels, 1)

    def forward(self, x, t, label):
        B, C, n_frames, J = x.shape
        x = x.view(B, -1, n_frames)

        t_vec = t.view(B, 1).float() / 1000.0
        t_emb = self.time_mlp(t_vec) + self.class_emb(label)

        x1 = self.inc(x)
        x2 = self.down1(x1, t_emb)
        x3 = self.down2(x2, t_emb)

        x3 = self.mid1(x3, t_emb)
        x3 = self.mid2(x3, t_emb)

        x_up1 = F.interpolate(x3, scale_factor=1.0, mode='nearest')

        x_up1 = torch.cat([x3, x3], dim=1)
        x4 = self.up1(x_up1, t_emb)

        x_up2 = torch.cat([x4, x2], dim=1)
        x5 = self.up2(x_up2, t_emb)

        out = self.outc(x5)
        return out.view(B, C, n_frames, J)

model = FullDiffusionModel(num_joints=JOINTS).to(DEVICE)
optimizer = torch.optim.Adam(model.parameters(), lr=2e-4)
loss_fn = nn.MSELoss()

class DiffusionManager:
    def __init__(self, timesteps=500, beta_start=1e-4, beta_end=0.02):
        self.timesteps = timesteps
        self.betas = torch.linspace(beta_start, beta_end, timesteps).to(DEVICE)
        self.alphas = 1. - self.betas
        self.alphas_cumprod = torch.cumprod(self.alphas, axis=0)
        self.sqrt_alphas_cumprod = torch.sqrt(self.alphas_cumprod)
        self.sqrt_one_minus_alphas_cumprod = torch.sqrt(1. - self.alphas_cumprod)

    def add_noise(self, x_start, t):
        noise = torch.randn_like(x_start)
        s1 = self.sqrt_alphas_cumprod[t].reshape(-1, 1, 1)
        s2 = self.sqrt_one_minus_alphas_cumprod[t].reshape(-1, 1, 1)
        return s1 * x_start + s2 * noise, noise

    def sample(self, model, n_samples, label, input_dim=45):
        model.eval()
        with torch.no_grad():
            x = torch.randn((n_samples, input_dim, FRAMES)).to(DEVICE)
            labels = torch.tensor([label] * n_samples).to(DEVICE)

            for i in reversed(range(self.timesteps)):
                t = torch.tensor([i] * n_samples, device=DEVICE).long()
                predicted_noise = model(x, t, labels)

                alpha = self.alphas[i]
                alpha_cumprod = self.alphas_cumprod[i]
                beta = self.betas[i]

                if i > 0:
                    noise = torch.randn_like(x)
                else:
                    noise = 0

                x = (1 / torch.sqrt(alpha)) * (x - ((1 - alpha) / (torch.sqrt(1 - alpha_cumprod))) * predicted_noise) + torch.sqrt(beta) * noise
        return x
```

```python
def calculate_connections(batch_motion):
    lengths = []
    for (start_idx, end_idx) in CONNECTIONS:
        p1 = batch_motion[..., start_idx]
        p2 = batch_motion[..., end_idx]

        dist = torch.norm(p1 - p2, dim=1)
        lengths.append(dist)

    return torch.stack(lengths, dim=2)

beta = torch.linspace(1e-4, 0.02, TIMESTEPS).to(DEVICE)
alpha = 1 - beta
alpha_bar = torch.cumprod(alpha, dim=0)

model.train()

W_MSE = 1.0
W_BONE = 0.5
print("Start")
for epoch in range(EPOCHS):
    total_loss = 0
    bone_loss_sum = 0

    for x, y in dataloader:
        x, y = x.to(DEVICE), y.to(DEVICE)

        t = torch.randint(0, TIMESTEPS, (x.size(0),), device=DEVICE)

        noise = torch.randn_like(x)
        a_bar_t = alpha_bar[t].view(-1, 1, 1, 1)

        x_noisy = torch.sqrt(a_bar_t) * x + torch.sqrt(1 - a_bar_t) * noise

        pred_noise = model(x_noisy, t, y)

        loss_mse = loss_fn(pred_noise, noise)
        pred_x0 = (x_noisy - torch.sqrt(1 - a_bar_t) * pred_noise) / torch.sqrt(a_bar_t)

        real_lengths = calculate_connections(x)
        pred_lengths = calculate_connections(pred_x0)
        loss_bone = F.mse_loss(pred_lengths, real_lengths)
        loss = (W_MSE * loss_mse) + (W_BONE * loss_bone)

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += loss.item()
        bone_loss_sum += loss_bone.item()

    if (epoch+1) % 200 == 0 or (epoch+1) < 10:
        avg_mse = total_loss / len(dataloader)
        avg_bone = bone_loss_sum / len(dataloader)
        print(f"Epoka {epoch+1}: Total Loss {avg_mse:.5f} | Bone Loss {avg_bone:.5f}")

print("Trening zakończony.")
```

```python
from scipy.signal import savgol_filter
import pandas as pd

def smooth_motion(motion_data):
    smoothed = np.zeros_like(motion_data)
    for j in range(motion_data.shape[2]):
        for c in range(3):
            try:
                smoothed[c, :, j] = savgol_filter(motion_data[c, :, j], window_length=9, polyorder=2)
            except:
                smoothed[c, :, j] = motion_data[c, :, j]
    return smoothed


@torch.no_grad()
def generate(action_name, do_smooth=True):
    model.eval()
    label = torch.tensor([0 if action_name=="walk" else 1], device=DEVICE)

    x = torch.randn(1, 3, FRAMES, JOINTS).to(DEVICE)

    for i in reversed(range(TIMESTEPS)):
        t = torch.tensor([i], device=DEVICE)

        pred_noise = model(x, t, label)
        if 'cosine_beta_schedule' in globals():
             curr_alpha = alpha[i]
             curr_alpha_bar = alpha_bar[i]
             curr_beta = beta[i]
        else:
             curr_beta = torch.linspace(1e-4, 0.02, TIMESTEPS).to(DEVICE)[i]
             curr_alpha = 1. - curr_beta
             curr_alpha_bar = torch.cumprod(torch.linspace(1e-4, 0.02, TIMESTEPS), 0).to(DEVICE)[i]
             curr_alpha_bar = alpha_bar[i]

        noise = torch.randn_like(x) if i > 0 else 0

        x = (1 / torch.sqrt(curr_alpha)) * (x - ((1 - curr_alpha) / (torch.sqrt(1 - curr_alpha_bar))) * pred_noise)
        x = x + torch.sqrt(curr_beta) * noise

    final_motion = x.cpu().numpy()[0]
    if do_smooth:
        final_motion = smooth_motion(final_motion)

    return final_motion


def plot_and_save_stickman(data, title, filename_base):
    fig = plt.figure(figsize=(5,5))
    ax = fig.add_subplot(111, projection='3d')
    connections = [(0,1),(1,2), (0,3),(3,4), (0,5),(5,6), (0,7),(7,8)]
    data = data.transpose(1, 2, 0)
    lines = [ax.plot([],[],[], 'o-', lw=2)[0] for _ in connections]

    def init():
        ax.set_xlim(-0.8, 0.8); ax.set_ylim(-0.8, 0.8); ax.set_zlim(-0.8, 0.8)
        ax.set_title(title)
        return lines

    def update(frame):
        pose = data[frame]
        for line, (s, e) in zip(lines, connections):
            line.set_data([pose[s,0], pose[e,0]], [pose[s,2], pose[e,2]])
            line.set_3d_properties([pose[s,1], pose[e,1]])
        return lines

    anim = animation.FuncAnimation(fig, update, frames=FRAMES, init_func=init, blit=False)

    gif_path = os.path.join(DATASET_ROOT, f"{filename_base}.gif")
    anim.save(gif_path, writer='pillow', fps=FRAMES/2)

    plt.close()
    return HTML(anim.to_jshtml())

anim_walk_data = generate("walk", do_smooth=True)
display(plot_and_save_stickman(anim_walk_data, "Walk", "animation_walk"))

anim_jump_data = generate("jump", do_smooth=True)
display(plot_and_save_stickman(anim_jump_data, "Jump", "animation_jump"))

```

```python
import numpy as np
from scipy import linalg
import torch
import numpy as np
from scipy import linalg


def calculate_mpjpe_proper(real, gen):
    """
    Liczy MPJPE (Mean Per Joint Position Error) poprawnie obsługując wymiary.
    Input: [N, 3, Frames, Joints] LUB [N, Frames, Joints, 3]
    """
    if real.shape[1] == 3:
        real = np.transpose(real, (0, 2, 3, 1))
        gen = np.transpose(gen, (0, 2, 3, 1))

    diff = real - gen

    dist = np.linalg.norm(diff, axis=-1)
    return np.mean(dist)

def calculate_fmd_raw(real_flat, gen_flat):
    mu1, sigma1 = np.mean(real_flat, axis=0), np.cov(real_flat, rowvar=False)
    mu2, sigma2 = np.mean(gen_flat, axis=0), np.cov(gen_flat, rowvar=False)
    diff = mu1 - mu2
    covmean, _ = linalg.sqrtm(sigma1.dot(sigma2), disp=False)
    if np.iscomplexobj(covmean): covmean = covmean.real
    return diff.dot(diff) + np.trace(sigma1) + np.trace(sigma2) - 2 * np.trace(covmean)

def calculate_diversity(gen):
    if gen.shape[1] == 3:
        gen = np.transpose(gen, (0, 2, 3, 1))
    N = gen.shape[0]
    gen_flat = gen.reshape(N, -1)

    if N < 2: return 0

    divs = []
    for _ in range(100):
        idx1, idx2 = np.random.randint(0, N, 2)
        dist = np.linalg.norm(gen_flat[idx1] - gen_flat[idx2])
        divs.append(dist)
    return np.mean(divs)

def get_real_data_from_loader(loader, device):
    """
    Przechodzi przez cały dataloader i zbiera wszystkie próbki do tablic NumPy.
    Rozdziela je według etykiet (0 = walk, 1 = jump).
    """
    real_data_dict = {"walk": [], "jump": []}


    for batch in loader:
        motions, labels = batch


        motions = motions.cpu().numpy()
        labels = labels.cpu().numpy()
        for i in range(len(labels)):
            if labels[i] == 0:
                real_data_dict["walk"].append(motions[i])
            elif labels[i] == 1:
                real_data_dict["jump"].append(motions[i])
    real_data_dict["walk"] = np.array(real_data_dict["walk"])
    real_data_dict["jump"] = np.array(real_data_dict["jump"])

    print(f"Znaleziono: {len(real_data_dict['walk'])} chodu, {len(real_data_dict['jump'])} skoków.")
    return real_data_dict

real_data_collection = get_real_data_from_loader(dataloader, DEVICE)
results = []

print(f"Rozpoczynam poprawioną ewaluację (24 klatki, 9 stawów)...")

for act in ["walk", "jump"]:
    real_batch = real_data_collection[act][:100]

    gen_list = []
    for _ in range(len(real_batch)):
        motion = generate(act, do_smooth=True)
        gen_list.append(motion)
    gen_batch = np.array(gen_list)

    mpjpe = calculate_mpjpe_proper(real_batch, gen_batch)

    real_flat = real_batch.reshape(real_batch.shape[0], -1)
    gen_flat = gen_batch.reshape(gen_batch.shape[0], -1)
    fmd = calculate_fmd_raw(real_flat, gen_flat)

    div = calculate_diversity(gen_batch)

    results.append({
        "Ruch": act,
        "FMD": round(fmd, 4),
        "MPJPE": round(mpjpe, 4)
        "Diversity": round(div, 4)
    })

import pandas as pd
df = pd.DataFrame(results)
display(df)
```
